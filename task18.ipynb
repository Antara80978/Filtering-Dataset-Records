```python
import pandas as pd

# Sample dataset matching a Student Performance structure
data = {
    'Student_ID': [101, 102, 103, 104, 105, 106, 107],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace'],
    'Age': [19, 21, 20, 22, 19, 21, 20],
    'Grade': ['A', 'B', 'A', 'C', 'B', 'A', 'F'],
    'Score': [92, 78, 88, 65, 82, 95, 45],
    'Passed': [True, True, True, True, True, True, False]
}

df = pd.DataFrame(data)
print("--- Original DataFrame ---")
print(df)
print(f"Total Records: {len(df)}\n")

# ==========================================
# Example 1: Single Boolean Condition
# ==========================================
# Filter students with a score greater than 80
high_scorers = df[df['Score'] > 80]

print("--- Example 1: Score > 80 ---")
print(high_scorers)
print(f"Resulting Records: {len(high_scorers)}\n")

# Explanation 1:
# The expression `df['Score'] > 80` produces a Series of True/False values.
# Passing this boolean mask to `df[...]` returns only the rows where the condition is True.


# ==========================================
# Example 2: Multiple Conditions with AND (&)
# ==========================================
# Filter students who are older than 19 AND have a Grade of 'A'
top_older_students = df[(df['Age'] > 19) & (df['Grade'] == 'A')]

print("--- Example 2: Age > 19 AND Grade == 'A' ---")
print(top_older_students)
print(f"Resulting Records: {len(top_older_students)}\n")

# Explanation 2:
# The bitwise AND operator `&` evaluates element-wise boolean operations.
# Each condition must be wrapped in parentheses `()` so Python evaluates
# the comparison operators (`>`, `==`) before the logical operator (`&`).


# ==========================================
# Example 3: Multiple Conditions with OR (|)
# ==========================================
# Filter students with a Score under 50 OR who failed
at_risk_students = df[(df['Score'] < 50) | (df['Passed'] == False)]

print("--- Example 3: Score < 50 OR Passed == False ---")
print(at_risk_students)
print(f"Resulting Records: {len(at_risk_students)}\n")

# Explanation 3:
# The bitwise OR operator `|` selects rows where either condition (or both) is True.
```
